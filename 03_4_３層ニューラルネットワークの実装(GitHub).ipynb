## 3.4.3 実装のまとめ

```python
# 図（３−１８）、図（３−１９）、図（３−２０）を実装する
# 数値は任意に決めた数字なので、気にしなくてよいが、次元や形は確認すること（計算エラーを起こさないように）
# 図→抽象的な式→そして具体的な式をイメージできるように

import numpy as np

# シグモイド関数を定義
def sigmoid(x):
  return 1/(1 + np.exp(-x))

# 恒等関数を定義
def identity_function(x):
  return x


# 関数を定義
def init_network():

# 辞書（本棚のイメージ）
  network = {}

  network['W1'] = np.array([[0.1,0.3,0.5],[0.2,0.4,0.6]])
  network['b1'] = np.array([0.1,0.2,0.3])
  network['W2'] = np.array([[0.1,0.4],[0.2,0.5],[0.3,0.6]])
  network['b2'] = np.array([0.1,0.2])
  network['W3'] = np.array([[0.1,0.3],[0.2,0.4]])
  network['b3'] = np.array([0.1,0.2])

  return network

def forward(network, x):
  W1, W2, W3 = network['W1'], network['W2'], network['W3']
  b1, b2, b3 = network['b1'], network['b2'], network['b3']

  a1 = np.dot(x, W1) + b1
  z1 = sigmoid(a1)

  a2 = np.dot(z1, W2) + b2
  z2 = sigmoid(a2)

  a3 = np.dot(z2, W3) + b3


  y = identity_function(a3)

  return y

network = init_network()   # returnで返ってくる値をしっかり受けnetwork変数に代入。それをそれをforward関数に渡す。
x = np.array([1.0,0.5])    # NumPy配列つくってxに代入。それをforward関数に渡す。
y = forward(network, x)    # forward関数で計算した値をしっかり受け取り、表示(print)する。
print(y)
```
