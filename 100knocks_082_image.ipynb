## 回答2
そう予想されるのは、ここまでの「次元が上がる（条件が追加される）」という本質的なイメージが完璧に頭に出来上がっている証拠です！<br>

------------------------------
## なぜ cap.shape では4次元の数字が出ないのか？
理由は、cap = cv2.VideoCapture() という命令が、動画データをすべてメモリ（パソコンの脳みそ）に載せているわけではなく、「動画ファイルを外から1コマずつストローで吸い上げるためのパイプ」のような役割しか持っていないからです。<br><br>
パイプ（cap）自体は「いま動画の中に何枚の画像があるか」の合計数（300枚）や、1枚ずつのサイズ（720×1280）というカタログスペック（設定値）は知っています。<br><br>
しかし、まだデータを吸い上げて合体させていないため、「4次元の塊」としてはまだこの世に存在していないのです。<br>そのため、.shape と聞かれても「まだ塊になってないから答えられない！」とエラーになってしまいます。

------------------------------
## あなたの予想（4次元のshape）を現実に確認するコード
では、動画をすべて吸い上げて、あなたの予想通り (300, 720, 1280, 3) という4次元の形（Tensor）を実際に作って確認してみましょう！<br><br>
Colabで以下のコードを新しいセルに貼り付けて実行してみてください。<br>動画の全フレームをNumPyの配列（塊）にギュッと詰め込むコードです。
<br><br>
【Python】<br>
import cv2<br>
import numpy as np<br>
cap = cv2.VideoCapture("mov/mov01.avi")<br>
frame_list = []<br>
<br>

＃　パイプから1コマずつ最後まで全て吸い上げる<br>

while cap.isOpened():<br>
　　ret, frame = cap.read()<br>
　　if not ret:<br>
　　　　break<br>

<br>

<br>
<br>
＃　溜まった画像のリストを、ガッチャンコして「4次元の塊」に変身させる！<br>
<br>
video_tensor = np.array(frame_list)<br><br>

＃　あなたの予想した4次元のshapeをプリント！<br><br>
