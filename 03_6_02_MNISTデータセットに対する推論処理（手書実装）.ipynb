```python
import sys,os
sys.path.append(os.pardir)
import numpy as np
from common.functions import sigmoid, softmax
from dataset.mnist import load_mnist
```

```python
# ③ダウンロードしたフォルダから学習済みファイルを読み込む（ピクル）

import pickle
def init_network():
  with open("sample_weight.pkl", 'rb') as f:
    network = pickle.load(f)

  return network
```

```python
# 関数１

def get_data():
  (x_train,t_train), (x_test, t_test) = load_mnist(normalize=True, flatten=True, one_hot_label=False)

  return x_test, t_test
```

```python
def predict(network, x):
  W1, W2, W3 = network['W1'], network['W2'], network['W3']
  b1, b2, b3 = network['b1'], network['b2'], network['b3']

  # 第1層の計算
  a1 = np.dot(x, W1) + b1
  z1 = sigmoid(a1) # シグモイド関数で変換

  # 第2層の計算
  a2 = np.dot(z1, W2) + b2
  z2 = sigmoid(a2) # シグモイド関数で変換

  # 第3層（出力層）の計算
  a3 = np.dot(z2, W3) + b3
  y = softmax(a3) # 最後にソフトマックス関数で確率にする

  return y
```

```python
network = init_network()  # 1. ネットワーク（重み）の読み込み
accuracy_cnt = 0 # 正解した数を数えるカウンターを設置

for i in range(len(x_test)):
  y = predict(network, x_test[i]) # 2. 1枚ずつ予測させる
  p = np.argmax(y)  # 一番確率が高い「数字（インデックス）」を取り出す

  if p == t_test[i]:
    accuracy_cnt += 1
```

```python
print("Accuracy:" + str(float(accuracy_cnt) / len(x_test)))
```
