結論から言うと、「1か0か（伝えるか伝えないか）」だけでなく、「量的なもの（どれくらいの強さで伝えるか）」があります。
質問者様が「1か0か」と理解されていたのは、まさに正しい歴史のスタートラインです。本の中で最初に登場するステップ関数（階段関数）は、まさに「1（伝える）か0（伝えないか）」の2択でした。
しかし、現代のディープラーニングでは、主に量的な強さを表す関数（シグモイド関数やReLU関数など）が使われています。なぜ量的な要素が必要なのか、分かりやすく解説します。

1. 1か0だけ（ステップ関数）の限界

初期の仕組み（パーセプトロン）では、信号の合計値が基準を超えたら「1（発火する）」、超えなければ「0（発火しない）」という仕組みでした。
しかし、これだと「あと一歩で基準に届きそう」という状態も「全くお話にならない状態」も、どちらも同じ「0」になってしまいます。これでは、AIが学習する（少しずつ賢くなる）ときに、「もう少しで正解だったよ」という微調整ができません。

2. 量的な伝える強さ（現代の活性化関数）

そこで登場するのが、本でも紹介されているシグモイド関数やReLU（レルー）関数です。これらは「量（グラデーション）」を伝えます。
• シグモイド関数：
出力を「0.0 〜 1.0」の間の連続した数値（実数）に変換します。
例えば、「0.2の強さで伝える」「0.85の強さで伝える」といったイメージです。これにより、「ギリギリ合格ラインを超えたから、0.51の強さで次に伝えよう」といった柔軟な表現が可能になります。
• ReLU関数：
現代の主流です。マイナスの入力は「0（伝えない）」ですが、プラスの入力は「増えれば増えるほど、その量（数値）をそのままストレートに伝える」という性質を持っています。

人間の脳のニューロンは、確かに「パルス（電気信号）を出すか出さないか（1か0か）」で動いていると言われます。
しかし、脳は「単位時間あたりに何回連続してパルスを出したか（頻度）」によって、情報の強さを量的に表現しています。ディープラーニングの活性化関数が「量」を出力するのは、この「信号の頻度（強さ）」を1つの数値で効率よく表現するためでもあります。
したがって、ご自身のまとめの「どれだけ伝えるか」という表現は、ディープラーニングの本質（グラデーションによる微調整）を捉えた完璧な表現です！自信を持って進めてください。

🗺️ 「分類」と「境界線」のイメージ

AIが写真を見て「犬か猫か」を判断したり、データを「グループAかグループBか」に分けることを「分類」と言います。

• 線形（直線）： まっすぐな定規で1本だけ線を引くこと
• 非線形（曲線・折れ曲がり）： ぐにゃぐにゃした線や、折れ曲がった線で囲むこと

🥦 なぜ「直線（活性化関数なし）」だとダメなのか？

ここに、「赤いトマト（甘い）」と「青いトマト（酸っぱい）」が散らばっている農場があるとします。
1. 簡単な農場（線形）： 左側に赤いトマト、右側に青いトマトが綺麗に分かれているなら、まっすぐな直線（定規）を1本引くだけで「甘い」と「酸っぱい」を綺麗に分類できます。活性化関数はいりません。
2. 複雑な農場（非線形）： もし、「農場の真ん中のエリアだけ、なぜか甘い赤いトマトが実る。その周りは全部青いトマト」という状態だったらどうでしょう？
まっすぐな定規（直線）をどう置いても、「真ん中の丸いエリア」だけを綺麗に囲むことはできませんよね。どれだけ定規を何本重ねて置いても、直線は直線のままなので、丸いエリアをくり抜くことはできません。

⚡ 活性化関数（ReLU）が入るとどうなる？

ここで、現代の主流であるReLU関数を思い出してみましょう。
ReLUは「マイナスは0、プラスはそのまま」という、「0のところでカクッと折れ曲がった関数」です。
この「カクッ」という折れ曲がり（非線形）がAIに加わると、魔法が起きます。
1. 1つのニューロンが「ここでカクッと折れ曲がった仕切り」を作ります。
2. 別のニューロンも「別の場所でカクッと折れ曲がった仕切り」を作ります。
3. これらを何層も重ね合わせることで、「カク、カク、カク、カク……」と、直線を折り曲げて、真ん中のエリアを囲む「多角形（のちに丸に近い形）」の境界線を作ることができるようになります。
このように、活性化関数がグラフを「折り曲げて（あるいはカーブさせて）」くれるおかげで、AIは複雑な形をした境界線を描き、「この境界線の内側なら犬、外側なら猫！」という複雑な分類ができるようになるのです。

まとめ

• 活性化関数がない： どんだけ層を重ねても「無限に長い1本のまっすぐな定規」しか使えない（複雑な形を囲めない）。
• 活性化関数がある： 定規を「折り曲げる」ことができる。何層も重ねることで、どんな複雑な形（境界線）でも作れるようになり、複雑なデータを正しく「分類」できるようになる。

🗺️ 「山と海」のイメージ

今、手元に「平らな1枚の紙」があるとします。
この紙の上には、データ（例えば、犬のデータと猫のデータ）が、あちこちに混ざって散らばっています。
1. 活性化関数がない場合（平らな紙のまま）：

2. 活性化関数がある場合（紙を折り曲げられる）：
ニューロンと活性化関数が計算を行うたびに、この平らな紙が「ぐにゃっ」と立体的に折り曲げられます。何層も重ねていくと、紙の「犬のデータがある場所」だけがグッと上に押し上げられて「山」になり、「猫のデータがある場所」はそのまま「低い土地（海）」になります。

• 水面から上に飛び出している「山」＝ 犬
• 水の中に沈んでいる「低い土地」＝ 猫

🧱 活性化関数に届くまでの1本道

計算の流れは、質問者様が整理された通りの順番で進みます。
1. 「入力 × 重み」の総和：

4. 活性化関数（ReLUなど）：

💡 なぜ「バイアス」を調整すると「折り曲げる位置」が変わるのか？

📈 「回帰（予測）」のビジュアルイメージ

分類では、「水没するかどうか（1か0か）」で境界線を作って仲間分けをしました。
一方、回帰は「明日の株価」や「来週の気温」のように、具体的な「数値（連続値）」そのものをピタリと当てるのが目的です。
回帰における「出力層」の役割（恒等関数）は、水を注ぐことではなく、「できあがったでこぼこ鉄板の『高さ』を、そのまま数値として読み取る」ことです。

🎢 回帰のビジュアル：ジェットコースターのレール

例えば、「過去の気温（入力）」から「明日のアイスの売上（出力）」を予測するAIを作るとします。
1. 隠れ層（中身）：
分類のときと全く同じです。重み、バイアス、活性化関数（金づち）を使って、鉄板をぐにゃぐにゃに変形させていきます。
2. 出力層（恒等関数）：
できあがった鉄板の曲面が、そのまま「予測値のグラフ（レール）」になります。
3. 予測の瞬間：
「明日の気温は30度」というデータを入力すると、鉄板の「30度」の場所をピンポイントで見ます。その場所の鉄板の高さが「50センチ」なら、売上予測は「50万円」、とそのまま高さを読み取るのです。
つまり、回帰とは「データの動き（波）にピッタリ重なるような、滑らかなでこぼこ鉄板（滑り台やレールの形）を作り出すこと」そのものです。

🔄 「分類」と「回帰」の違いのまとめ

質問者様が気づかれた通り、途中の「空間を歪める（でこぼこ鉄板を作る）」プロセスは完全に同じです。最後の出力層だけが異なります。

項目	分類（ソフトマックス関数など）	回帰（恒等関数）
やりたいこと	データをグループに分ける	具体的な数値をピタリと当てる
出力層の役割	鉄板に水を注いで、水面でスパッと仕切る	鉄板の「高さ」をそのまま定規で測る
ビジュアル	水たまりの「境界線」	鉄板そのものの「カーブ（レール）」

本の中で「分類にはソフトマックス、回帰には恒等関数」と書かれているのは、まさにこの「水を注いで境界線を作るか」か「高さをそのまま測るか」の違いを、計算式（関数）として切り替えているだけなのです。
